## Noi dung bai
Sinh 1000 diem du lieu theo hinh tron tam O(0,0) ban kinh R trong khoang [0, 1]
- X: chieu truc hoanh [-1, 1]
- Y: chieu truc tung [-1, 1]
## Bai lam

### Import thu vien

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Sinh ngau nhien hinh tron

```python
# generate 1000 random points in circle
r = np.random.uniform(0, 1, 1000)
theta = np.random.uniform(0, 2 * np.pi, 1000)
x = r * np.cos(theta)
y = r * np.sin(theta)
```

```python
print(f'+ x: |x|={len(x)}, x in [{np.min(x):.2f}, {np.max(x):.2f}], 10 samples: {x[0:10]}')
print(f'+ y: |y|={len(y)}, y in [{np.min(y):.2f}, {np.max(y):.2f}], 10 samples: {y[0:10]}')
```

### Hien thi do thi

```python
# visualize data
plt.figure(figsize=(6, 6))
plt.scatter(x, y, color="blue", s=5)
plt.xlim(-1.1, 1.1)
plt.ylim(-1.1, 1.1)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("1000 random points in circle")
plt.grid(True)
plt.show()
```
