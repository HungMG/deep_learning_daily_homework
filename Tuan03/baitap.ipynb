# BÀI TẬP THỰC HÀNH TUẦN 4 - TRỰC QUAN HÓA DỮ LIỆU

**Chủ đề:** Trực quan hóa dữ liệu thống kê (Histogram, Boxplot, Scatter Plot) & Phát hiện giá trị dị biệt (Outliers/Errors).

- **Công cụ:** Python, Matplotlib, Seaborn, Pandas, Numpy
- **Mục tiêu:** Thực hành vẽ biểu đồ tần số, biểu đồ hộp, biểu đồ phân tán để phân tích dữ liệu và nhận diện ngoại lai.

## 0. Cấu hình môi trường và nạp thư viện
Cài đặt cấu hình hiển thị đồ thị và các thư viện cần thiết (`pandas`, `numpy`, `matplotlib`, `seaborn`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cấu hình giao diện tiếng Việt và đồ họa sạch sẽ
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 120
```

---
## 1. Hình 1: Histogram phân phối cột Tuổi (Câu 1 - Q3)
- **Tập dữ liệu**: Gồm 9 mẫu tuổi hợp lệ: `[25, 30, 22, 40, 35, 29, 50, 27, 31]`.
- **Các khoảng phân lớp (bins)**: `[20-30)`, `[30-40)`, `[40-50]`.
- **Ý nghĩa**: Đếm số lượng mẫu rơi vào từng nhóm tuổi và hiển thị nhãn tần số trực tiếp trên các cột.

```python
# 1. HISTOGRAM CỘT TUỔI (CÂU 1 - Q3)
tuoi_c1 = [25, 30, 22, 40, 35, 29, 50, 27, 31]  # 9 mẫu hợp lệ

fig, ax = plt.subplots(figsize=(7, 4.5))

# Định nghĩa các khoảng (bins): [20-30), [30-40), [40-50]
n, bins, patches = ax.hist(
    tuoi_c1,
    bins=[20, 30, 40, 50],
    edgecolor='black',
    color='#2b6cb0',
    alpha=0.85,
    rwidth=1.0
)

# Hiển thị số lượng mẫu trên từng cột
for i in range(len(patches)):
    if n[i] > 0:
        ax.text(
            patches[i].get_x() + patches[i].get_width() / 2,
            n[i] + 0.15,
            f'{int(n[i])}',
            ha='center',
            va='bottom',
            fontweight='bold',
            color='#1a365d',
            fontsize=11
        )

ax.set_title("1. HISTOGRAM PHÂN PHỐI CỘT TUỔI (CÂU 1)", fontsize=12, fontweight='bold', pad=12, color='#1a365d')
ax.set_xlabel("Khoảng Tuổi (năm)", fontweight='bold', labelpad=8)
ax.set_ylabel("Số lượng mẫu (Tần số)", fontweight='bold', labelpad=8)
ax.set_xticks([20, 30, 40, 50])
ax.set_ylim(0, 5)
plt.tight_layout()
plt.show()
```

---
## 2. Hình 2: Boxplot cột Lương & Outliers (Câu 1 - Q3/Q8)
- **Tập dữ liệu**: Mức lương gồm 9 mẫu: `[8000, 5000, 15000, -2000, 7000, 8500, 30000, 8200, 7800]`.
- **Các mốc thống kê quan trọng**:
  - $Q_1 = 7,400$
  - $\text{Median} = 8,000$
  - $Q_3 = 11,750$
  - **Outlier**: `30,000` (giá trị cao bất thường so với phân phối)
  - **Dữ liệu lỗi**: `-2,000` (mức lương âm không hợp lệ trong thực tế)

```python
# 2. BOXPLOT CỘT LƯƠNG (CÂU 1 - Q3/Q8)
luong_c1 = [8000, 5000, 15000, -2000, 7000, 8500, 30000, 8200, 7800]

fig, ax = plt.subplots(figsize=(6, 5))
box = ax.boxplot(
    luong_c1,
    vert=True,
    patch_artist=True,
    boxprops=dict(facecolor='#ebf8ff', color='#2b6cb0', linewidth=1.5),
    capprops=dict(color='#2b6cb0', linewidth=1.5),
    whiskerprops=dict(color='#2b6cb0', linewidth=1.5),
    flierprops=dict(marker='o', markerfacecolor='red', markersize=8, linestyle='none'),
    medianprops=dict(color='red', linewidth=2)
)

# Chú thích các mốc quan trọng
q1, median, q3 = 7400, 8000, 11750
ax.text(1.15, q1, f'Q1 = {q1}', va='center', fontsize=9.5, color='#2d3748')
ax.text(1.15, median, f'Median = {median}', va='center', fontsize=9.5, color='red', fontweight='bold')
ax.text(1.15, q3, f'Q3 = {q3}', va='center', fontsize=9.5, color='#2d3748')
ax.text(1.15, 30000, 'Outlier (30,000)', va='center', fontsize=9.5, color='red', fontweight='bold')
ax.text(1.15, -2000, 'Lỗi (-2,000)', va='center', fontsize=9.5, color='darkred', fontweight='bold')

ax.set_title("2. BOXPLOT BÁO CÁO LƯƠNG & OUTLIERS (CÂU 1)", fontsize=12, fontweight='bold', pad=12, color='#1a365d')
ax.set_ylabel("Mức Lương (VNĐ)", fontweight='bold', labelpad=8)
plt.tight_layout()
plt.show()
```

---
## 3. Hình 3: Scatter Plot tương quan Tuổi vs Lương (Câu 1 - Q4)
- **Mục tiêu**: Khảo sát mối quan hệ giữa biến Độ tuổi và Lương.
- **Nhận xét trực quan**:
  - Đánh dấu điểm lỗi lương âm: `Tuổi = 35, Lương = -2000`.
  - Đánh dấu điểm ngoại lai lương đột biến: `Tuổi = 50, Lương = 30000`.
  - Đường tham chiếu `y = 0` (đường nét đứt) giúp dễ dàng nhận biết ranh giới giá trị âm.

```python
# 3. SCATTER PLOT TUỔI VS LƯƠNG (CÂU 1 - Q4)
tuoi_scatter = [25, 22, 40, 35, 29, 50, 31]
luong_scatter = [8000, 5000, 15000, -2000, 8500, 30000, 7800]

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.scatter(tuoi_scatter, luong_scatter, color='#2b6cb0', s=80, edgecolors='black', zorder=3)

# Đánh dấu chú thích điểm lỗi và dị biệt
ax.annotate(
    'Lỗi: Lương âm (-2000)',
    xy=(35, -2000),
    xytext=(36, 2000),
    arrowprops=dict(facecolor='red', shrink=0.08, width=1, headwidth=5),
    fontsize=9,
    color='red',
    fontweight='bold'
)
ax.annotate(
    'Outlier: Lương cao (30000)',
    xy=(50, 30000),
    xytext=(38, 28000),
    arrowprops=dict(facecolor='red', shrink=0.08, width=1, headwidth=5),
    fontsize=9,
    color='red',
    fontweight='bold'
)
ax.axhline(0, color='gray', linestyle='--', linewidth=1, alpha=0.7)

ax.set_title("3. SCATTER PLOT TƯƠNG QUAN TUỔI VS LƯƠNG (CÂU 1)", fontsize=12, fontweight='bold', pad=12, color='#1a365d')
ax.set_xlabel("Tuổi (năm)", fontweight='bold', labelpad=8)
ax.set_ylabel("Lương (VNĐ)", fontweight='bold', labelpad=8)
plt.tight_layout()
plt.show()
```

## 4. Hình 4: Boxplot tập 27 giá trị Tuổi (Câu 2c)
- **Tập dữ liệu**: Gồm 27 giá trị tuổi đã sắp xếp thứ tự:
  `[13, 15, 16, 16, 19, 20, 20, 21, 22, 22, 25, 25, 25, 25, 30, 33, 33, 35, 35, 35, 35, 36, 40, 45, 46, 52, 70]`
- **Bộ 5 số tóm tắt (Five-number summary)**:
  - $\text{Min} = 13$
  - $Q_1 = 20$
  - $\text{Median} = 25$
  - $Q_3 = 35$
  - **Outlier**: `70` (nằm ngoài phạm vi $[Q_1 - 1.5 \times IQR, Q_3 + 1.5 \times IQR]$)

```python
# 4. BOXPLOT TẬP 27 GIÁ TRỊ TUỔI (CÂU 2c)
tuoi_c2 = [13, 15, 16, 16, 19, 20, 20, 21, 22, 22, 25, 25, 25, 25, 30, 33, 33, 35, 35, 35, 35, 36, 40, 45, 46, 52, 70]

fig, ax = plt.subplots(figsize=(6, 5))
ax.boxplot(
    tuoi_c2,
    vert=True,
    patch_artist=True,
    boxprops=dict(facecolor='#feebc8', color='#dd6b20', linewidth=1.5),
    capprops=dict(color='#dd6b20', linewidth=1.5),
    whiskerprops=dict(color='#dd6b20', linewidth=1.5),
    flierprops=dict(marker='o', markerfacecolor='red', markersize=8, linestyle='none'),
    medianprops=dict(color='red', linewidth=2)
)

# Chú thích 5 mốc thống kê
ax.text(1.15, 13, 'Min = 13', va='center', fontsize=9.5, color='#2d3748')
ax.text(1.15, 20, 'Q1 = 20', va='center', fontsize=9.5, color='#2d3748')
ax.text(1.15, 25, 'Median = 25', va='center', fontsize=9.5, color='red', fontweight='bold')
ax.text(1.15, 35, 'Q3 = 35', va='center', fontsize=9.5, color='#2d3748')
ax.text(1.15, 70, 'Outlier (70)', va='center', fontsize=9.5, color='red', fontweight='bold')

ax.set_title("4. BOXPLOT TẬP 27 GIÁ TRỊ TUỔI (CÂU 2c)", fontsize=12, fontweight='bold', pad=12, color='#1a365d')
ax.set_ylabel("Giá trị Tuổi", fontweight='bold', labelpad=8)
plt.tight_layout()
plt.show()
```
