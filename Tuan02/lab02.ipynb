## Bài 5. Reshape và Flatten

**Yêu cầu:**
- Kiểm tra số phần tử trước và sau khi reshape.
- Thử reshape Tensor thành kích thước 2 × 6.

```python
import torch

torch.manual_seed(42)
a = torch.randn(3, 4)

print("Original:")
print("shape =", a.shape, "| numel =", a.numel())
print(a)

r = a.ravel()
print("\nFlatten (ravel):")
print("shape =", r.shape, "| numel =", r.numel())
print(r)

b = a.reshape(3, 2, 2)
print("\nReshape to 3 x 2 x 2:")
print("shape =", b.shape, "| numel =", b.numel())

c = a.reshape(2, 6)   # Yêu cầu
print("\nReshape to 2 x 6:")
print("shape =", c.shape, "| numel =", c.numel())
print(c)
```

## Bài 6. Phép toán và nhân ma trận

**Yêu cầu:**
- Phân biệt phép toán `a * b` với `a @ b.T`.
- Cho biết shape của kết quả `a @ b.T`.

```python
import torch

torch.manual_seed(42)
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\na * b (phép toán phần tử, element-wise):")
print("shape =", (a * b).shape)
print(a * b)

print("\nMatrix multiplication a @ b.T:")
print("shape =", (a @ b.T).shape)
print(a @ b.T)
```

**Nhận xét:**
- `a * b`: **nhân từng phần tử** cùng vị trí (element-wise), shape giữ nguyên **(2, 3)**.
- `a @ b.T`: **nhân ma trận** — `(2,3) @ (3,2)`, phần tử `[i, j]` = tích vô hướng (dot product) hàng `i` của `a` với hàng `j` của `b`.
- Shape kết quả `a @ b.T` là **(2, 2)**.

**Nhận xét:**- `a * b`: **phép nhân từng phần tử** (element-wise) — nhân từng phần tử tương ứng, giữ nguyên shape `(2, 3)`.- `a @ b.T`: **phép nhân ma trận** — `(2, 3) @ (3, 2)`, phần tử `[i, j]` là tích vô hướng hàng `i` của `a` và hàng `j` của `b`.- Shape kết quả `a @ b.T` là **(2, 2)**.

## Bài 7. Các hàm thống kê trên Tensor

**Yêu cầu:** Thay `dim=0` bằng `dim=1` và nhận xét sự khác nhau.

```python
import torch

torch.manual_seed(42)
a = torch.randn(3, 4)

print(a)
print("\n--- dim=0 (theo cột) ---")
print("Mean shape:", torch.mean(a, dim=0).shape)
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))

print("\n--- dim=1 (theo hàng) ---")
print("Mean shape:", torch.mean(a, dim=1).shape)
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

**Nhận xét:**
- `dim=0`: giảm theo **chiều 0 (xác định qua các hàng)** → tính đại lượng trên từng **cột**, kết quả có shape `(4,)`.
- `dim=1`: giảm theo **chiều 1** → tính trên từng **hàng**, kết quả có shape `(3,)`.
- `cumsum(dim=0)` cộng dồn **xuống theo cột** (xây dựng tổng dọc), `cumsum(dim=1)` cộng dồn **sang theo hàng` (tổng tích lũy ngang).

## Bài 8. Autograd cơ bản

**Yêu cầu:**
- Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với `x.grad`.
- Đổi hàm thành `y = 3*x**2 + 2*x + 1` và kiểm tra gradient.

```python
import torch

# (a) y = x^2 tại x = 3.6
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()
print("(a) y = x^2")
print("x     =", x.item())
print("y     =", y.item())
print("x.grad =", x.grad.item())
print("Thủ công: dy/dx = 2x = 2 * 3.6 =", 2 * 3.6)

# (b) y = 3x^2 + 2x + 1
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2 ** 2 + 2 * x2 + 1
y2.backward()
print("\n(b) y = 3x^2 + 2x + 1")
print("x.grad =", x2.grad.item())
print("Thủ công: dy/dx = 6x + 2 =", 6 * 3.6 + 2)
```

**Nhận xét:**
- (a) Đạo hàm y = x² là `dy/dx = 2x` → tại x = 3.6, giá trị = **7.2**, khớp chính xác với `x.grad`.
- (b) Với y = 3x² + 2x + 1, `dy/dx = 6x + 2` → tại x = 3.6, giá trị = **23.6**, khớp `x.grad`.
→ Autograd tự động tính gradient đúng bằng chuỗi quy tắc đạo hàm, không cần tính tay.

## Bài 9. Gradient Descent tìm hệ số đa thức

**Yêu cầu:**
- So sánh ba hệ số học được với `[1, 2, 3]`.
- Giải thích vai trò của `optimizer.zero_grad()`, `mse.backward()` và `optimizer.step()`.

```python
import numpy as np
import torch

np.random.seed(42)
torch.manual_seed(42)

polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w.detach().flatten().tolist())

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w.detach().flatten().tolist())
print("Target: [1, 2, 3]")
print("Final MSE:", mse.item())
```

**Nhận xét:**
- Ba hệ số học được xấp xỉ **[1, 2, 3]** (sai số rất nhỏ sau 1000 epoch NAdam) → Gradient Descent đã khôi phục đúng hệ số đa thức y = x² + 2x + 3.
- Vai trò ba bước trong mỗi epoch:
  - `optimizer.zero_grad()`: **xóa gradient tích lũy** của epoch trước (nếu không xóa, gradient sẽ bị cộng dồn sai giữa các vòng lặp).
  - `mse.backward()`: autograd **tính gradient** ∂MSE/∂w rồi lưu vào `w.grad`.
  - `optimizer.step()`: optimizer (NAdam) **dùng gradient vừa tính để cập nhật** `w` theo hướng giảm MSE.

## Bài 10. Giải hệ phương trình bằng Autograd

**Yêu cầu:**
- Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C−D=1, A+C=8, B−D=2.
- Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
import random
import torch

torch.manual_seed(42)
random.seed(42)

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1 * y1 + y2 * y2 + y3 * y3 + y4 * y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())

print("\nKiểm tra 4 phương trình:")
print(f"A + B = {A.item() + B.item():.4f}  (cần 9)")
print(f"C - D = {C.item() - D.item():.4f}  (cần 1)")
print(f"A + C = {A.item() + C.item():.4f}  (cần 8)")
print(f"B - D = {A.item() * 0 + B.item() - D.item():.4f}  (cần 2)")
print(f"sqerr (tổng bình phương sai số) = {sqerr.item():.6f}")
```

**Nhận xét:**
- Nghiệm thu được thỏa cả 4 phương trình (sai số ≈ 0, `sqerr` rất nhỏ).
- **Vì sao có nhiều nghiệm:** Hệ có 4 ẩn nhưng chỉ có **rank = 3** — phương trình thứ 4 phụ thuộc tuyến tính 3 phương trình còn lại (cụ thể: `(A+B) − (B−D) − (A+C−8) ...` tạo ra tổ hợp bị phụ thuộc, hoặc kiểm tra: `eq1 − eq4` … xem thêm bên dưới). Ma trận hệ có một hạng tử tự do → nghiệm phụ thuộc **một tham số tự do** `t`:
  - `A = t`
  - `B = 9 − t`
  - `C = 8 − t`
  - `D = 7 − t` (thỏa `B − D = (9−t) − (7−t) = 2`)
→ Gradient Descent chỉ hội tụ về **một** nghiệm tùy điểm khởi tạo ngẫu nhiên, nên kết quả mỗi lần chạy có thể khác nhau nhưng đều thỏa hệ.

### 11.4. Đa thức y = 2x² − 3x + 5 (thay Bài 9), tìm hệ số

```python
import numpy as np
import torch

np.random.seed(42)
torch.manual_seed(42)

polynomial = np.poly1d([2, -3, 5])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()
    mse = torch.mean(torch.square(y - x @ w))
    mse.backward()
    optimizer.step()

print("Learned coefficients:", w.detach().flatten().tolist())
print("Target: [2, -3, 5]")
print("Final MSE:", mse.item())
```

**Nhận xét:** Hệ số học được xấp xỉ **[2, −3, 5]** → NAdam + autograd hội tụ đúng hệ số đa thức y = 2x² − 3x + 5.
